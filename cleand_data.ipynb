```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Tất cả thư viện hoạt động!")
```

```python
# --- BƯỚC 1: Đọc file ---
df = pd.read_csv('Superstore.csv')

print("="*50)
print("TỔNG QUAN DATASET")
print("="*50)
print(f"Số dòng (rows):      {df.shape[0]}")
print(f"Số cột (columns):    {df.shape[1]}")

print("\n" + "="*50)
print("KIỂU DỮ LIỆU TỪNG CỘT")
print("="*50)
print(df.dtypes)

print("\n" + "="*50)
print("5 DÒNG ĐẦU TIÊN")
print("="*50)
df.head()
```

```python
# --- BƯỚC 2: Kiểm tra dữ liệu thiếu và trùng lặp ---
print("GIÁ TRỊ THIẾU (NULL) TRONG MỖI CỘT:")
print(df.isnull().sum())
print(f"\nTổng giá trị null: {df.isnull().sum().sum()}")

print("\n" + "="*50)
print(f"Số dòng trùng lặp: {df.duplicated().sum()}")
```

```python
# --- BƯỚC 3: Chuyển đổi kiểu dữ liệu phù hợp ---

# Postal Code nên là string (không phải số)
df['Postal Code'] = df['Postal Code'].astype(str)

# Chuyển cột ngày sang datetime
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d-%m-%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d-%m-%Y')

print("Kiểu dữ liệu SAU khi chuyển đổi:")
print(df.dtypes)

print("\nKích thước dataframe:", df.shape)
print("\n5 dòng đầu tiên:")
df.head()
```
